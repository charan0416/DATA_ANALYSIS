```python
import pandas as pd
import numpy as np
import seaborn as sns
import plotly.express as px
import matplotlib.pyplot as plt
import plotly.graph_objects as go
```

```python
data = pd.read_csv('/content/Chocolate Sales.csv')
```

```python
most_sold = data.groupby('Product')['Boxes Shipped'].sum().sort_values(ascending=False)
```

```python
most_sold
```

```python
import seaborn as sns
import matplotlib.pyplot as plt


fig, ax = plt.subplots(figsize=(12, 6))
sns.barplot(x=most_sold.index, y=most_sold.values, ax=ax)


ax.set_xlabel("Product", fontsize=12)
ax.set_ylabel("Boxes Shipped", fontsize=12)
ax.set_title("Total Boxes Shipped by Product", fontsize=14)
plt.xticks(rotation=45, ha="right", fontsize=10)
plt.tight_layout()



```

```python
product_sales = data.groupby('Product')['Amount'].sum().sort_values(ascending=False).head(15)
sns.barplot(x=product_sales.values, y=product_sales.index, palette='magma')
plt.title('Top 15 Products by Total Sales')
plt.xlabel('Sales Amount ($)')
plt.ylabel('Product')
plt.tight_layout()
plt.show()
```

```python
data['Date'] = pd.to_datetime(data['Date'], format='%m/%d/%Y')
```

```python
data['Month'] = data['Date'].dt.to_period('M').astype(str)
   # ... (rest of your plotting code)
```

```python
data['Month'] = data['Date'].dt.to_period('M').astype(str)
monthly_sales = data.groupby('Month')['Amount'].sum()
monthly_sales.plot(kind='line', marker='o', color='teal')
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales ($)')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
sns.scatterplot(data=data, x='Boxes Shipped', y='Amount', hue='Country', alpha=0.7)
plt.title('Boxes Shipped vs. Sales Amount')
plt.xlabel('Boxes Shipped')
plt.ylabel('Sales Amount ($)')
plt.tight_layout()
plt.show()
```

```python
top_days =data.groupby('Date')['Amount'].sum().sort_values(ascending=False).head(10)
top_days.plot(kind='bar', color='blue')
plt.title('Top 10 Highest Sales Days')
plt.xlabel('Date')
plt.ylabel('Total Sales ($)')
plt.tight_layout()
plt.show()
```

```python
avg_product_sales = data.groupby('Product')['Amount'].mean().sort_values(ascending=False).head(10)
sns.barplot(x=avg_product_sales.values, y=avg_product_sales.index, palette='spring')
plt.title('Top 10 Products by Average Sale Amount')
plt.xlabel('Average Sale Amount ($)')
plt.ylabel('Product')
plt.tight_layout()
plt.show()

```

```python
sns.histplot(data['Amount'], bins=30, kde=True, color='purple')
plt.title('Distribution of Sales Amounts')
plt.xlabel('Sales Amount ($)')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
sns.boxplot(data=data, x='Country', y='Amount', palette='pastel')
plt.title('Sales Amount Distribution by Country')
plt.xlabel('Country')
plt.ylabel('Sales Amount ($)')
plt.tight_layout()
plt.show()
```

```python
heatmap_data = data.groupby(['Month', 'Country'])['Amount'].sum().unstack().fillna(0)
sns.heatmap(heatmap_data, annot=True, fmt=".0f", cmap="YlGnBu")
plt.title('Monthly Sales Heatmap by Country')
plt.xlabel('Country')
plt.ylabel('Month')
plt.tight_layout()
plt.show()
```
