```python
data1['day_of_week'] = data1['trip_start_date'].dt.day_name()

plt.figure(figsize=(10,6))
sns.barplot(x='day_of_week', y='delay_time', data=data1, order=[
    'Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday'])
plt.title('Delays by Day of the Week')
plt.xticks(rotation=100)
plt.show()
```

```python
data1['start_hour'] = data1['trip_start_date'].dt.hour

plt.figure(figsize=(10,6))
sns.barplot(x='start_hour', y='delay_time', data=data1)
plt.title('Delays by Start Hour of the Day')
plt.xlabel('Hour of the Day')
plt.ylabel('Delay (minutes)')
plt.grid(True)
plt.show()
```

```python
route_delay_avg = data1.groupby('route')['delay_time'].mean().sort_values()
print(route_delay_avg.head(10))
print(route_delay_avg.tail(10))
```

```python
vehicle_trip_counts = data1['vehicle_no'].value_counts().head(10)

plt.figure(figsize=(10,6))
vehicle_trip_counts.plot(kind='bar')
plt.title('Top 10 Most Active Vehicles (Trip Count)')
plt.ylabel('Number of Trips')
plt.xticks(rotation=45)
plt.show()
```

```python
data1['booking_to_start_hours'] = (data1['trip_start_date'] - data1['BookingID_Date']).dt.total_seconds() / 3600

plt.figure(figsize=(10,6))
sns.histplot(data1['booking_to_start_hours'].dropna(), bins=50, kde=True)
plt.title('Time Between Booking and Trip Start')
plt.xlabel('Hours')
plt.ylabel('Number of Trips')
plt.show()
```
